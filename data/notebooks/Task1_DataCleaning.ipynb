```python
import pandas as pd

df = pd.read_csv("retail_smart_store.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
df.isnull().sum()

df['Price'] = df['Price'].fillna(df['Price'].mean())
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['Purchase_Date'] = pd.to_datetime(df['Purchase_Date'])
```

```python
df['Total_Amount'] = df['Price'] * df['Quantity']
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(x='Product_Category', y='Total_Amount', data=df)
plt.title("Sales by Category")
plt.show()
```

```python
sns.barplot(x='City', y='Total_Amount', data=df)
plt.xticks(rotation=45)
plt.title("Sales by City")
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.boxplot(x=df['Total_Amount'])
plt.title("Outlier Detection")
plt.show()
```

## 📊 Key Insights

- Electronics category generates the highest revenue
- Bangalore and Mumbai have the highest customer spending
- Missing values in Price were handled using mean
- Duplicate records were removed to ensure accuracy
- Discounts slightly increase total purchase amount
- Outliers were identified using boxplot
