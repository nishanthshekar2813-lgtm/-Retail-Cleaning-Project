```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("retail_smart_store.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
plt.hist(df['Age'])
plt.title("Age Distribution")
plt.show()
```

```python
sns.countplot(x='Gender', data=df)
plt.title("Gender Distribution")
plt.show()
```

```python
sns.countplot(x='Product_Category', data=df)
plt.title("Product Category Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.scatter(df['Price'], df['Quantity'])
plt.title("Price vs Quantity")
plt.xlabel("Price")
plt.ylabel("Quantity")
plt.show()
```

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
df['Total_Amount'] = df['Price'] * df['Quantity']

df['High_Spender'] = df['Total_Amount'].apply(lambda x: 1 if x > 10000 else 0)
```

```python
sns.countplot(x='High_Spender', data=df)
plt.title("High Spender Distribution")
plt.show()
sns.boxplot(x='High_Spender', y='Total_Amount', data=df)
plt.show()
```

```python
sns.barplot(x='City', y='Total_Amount', data=df)
plt.xticks(rotation=45)
plt.title("City vs Total Spending")
plt.show()
```

📌 Insights
Most customers are between age 20–40
Electronics category has highest sales
High spenders are fewer but contribute more revenue
Strong correlation between Price and Total_Amount
Cities like Bangalore and Mumbai show higher spending

📊 Conclusion
Data shows clear customer segmentation
High spenders can be targeted for premium products
Product categories influence spending behavior
Business can use this for marketing strategies
