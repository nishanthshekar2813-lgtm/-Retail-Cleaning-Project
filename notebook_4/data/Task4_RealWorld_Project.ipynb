```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("online_retail.csv")
df.head()
```

```python
df.dropna(subset=['CustomerID'], inplace=True)
df = df[df['Quantity'] > 0]

df['Total_Amount'] = df['Quantity'] * df['UnitPrice']
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['Month'] = df['InvoiceDate'].dt.month

df['High_Spender'] = df['Total_Amount'].apply(lambda x: 1 if x > 500 else 0)
```

```python
df['Country'].value_counts().head(10).plot(kind='bar')
plt.title("Top Countries")
plt.show()
```

```python
df.groupby('Month')['Total_Amount'].sum().plot()
plt.title("Monthly Sales Trend")
plt.show()
```

```python
df.groupby('Description')['Quantity'].sum().sort_values(ascending=False).head(10).plot(kind='bar')
plt.title("Top Products")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Country'] = le.fit_transform(df['Country'])

X = df[['Quantity', 'UnitPrice', 'Country']]
y = df['High_Spender']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

dt = DecisionTreeClassifier()
rf = RandomForestClassifier()

dt.fit(X_train, y_train)
rf.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

print("Decision Tree:", accuracy_score(y_test, dt.predict(X_test)))
print("Random Forest:", accuracy_score(y_test, rf.predict(X_test)))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, rf.predict(X_test))

sns.heatmap(cm, annot=True)
plt.title("Confusion Matrix")
plt.show()
```
